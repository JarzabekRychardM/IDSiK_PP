## Zasady

---
### Punktacja i przeliczenia
- W każdym module zbieramy punkty
- Łącznie w kursie jest do zdobycia **50 punktów**:
  - Kartkówki + aktywność: do 40 punktów  
  - Kolokwium zaliczeniowe: 10 punktów
- Podczas każdych ćwiczeń wykonujemy zadania z udostępnionej listy zadań
- **Wymagane jest przygotowanie do wykonywania zadań z listy.**
- Każda lista zadań jest oddawana na zaliczenie (warunek dopuszczenia do kartkówki), ale sama lista nie generuje punktów
- Oddanie pracy (lub jej cześci) będącej plagiatem, skutkuje niezaliczeniem całego modułu
- **Zgłoszenie się, poprawne wykonanie i wyjaśnienie zadania** podczas ćwiczeń = **0.5 punkta** w ramach danego modułu
- Brak przygotowania do wykonania któregoś z zadań prowadzi do obniżenia maksymalnej ilości punktów możliwych do zdobycia w module o 0.5 punkta
- Punkty zdobyte na ćwiczeniach przechodzą na “konto” punktów z modułu, więc można zrobić mniej zadań na kartkówce (nadwyżka punktów nie przechodzi dalej, dlatego warto systematycznie przychodzić na zajęcia i utrzymywać aktywność)

### Przykład
Adrian na pierwszych zajęciach z Modułu 3 dwa razy się zgłosił i zdobył 1 punkt za poprawne wykonanie zadania. Na drugich zajęciach z tego modułu również poprawnie rozwiązał zadania z listy i zdobył kolejny punkt.  
Łącznie ma więc 2 punkty z Modułu 3. Podczas kartkówki, która została zapowiedziana z tego Modułu, na której do zdobycia będzie 5 punktów, Adrian będzie musiał zdobyć wyłącznie 3 punkty, aby uzbierać maksymalną ilość z Modułu, czyli 5. Brawo Adrian!


Ocena końcowa z kursu wystawiana jest na podstawie procentowej wartości punktów zdobytych w stosunku do wszystkich możliwych do uzyskania punktów (50):
- 51 - 60% -> 3.0
- 61 - 70% -> 3.5
- 71 - 80% -> 4.0
- 81 - 90% -> 4.5
-  91 -100% -> 5.0


### Przygotowanie do ćwiczeń
- przed zajęciami przeczytaj opis listy, skonfiguruj środowisko i pobierz dane; oddanie listy w terminie jest warunkiem podejścia do kartkówki;
- zgłoszenie się i obronienie rozwiązania przy tablicy = 0.5 pkt do konta modułu (liczone jako punkty aktywności);
- brak przygotowania obniża maksymalny wynik modułu o 0.5 pkt – jeśli masz problem techniczny, zgłoś to wcześniej prowadzącemu.

Checklist przed każdą sesją: (1) środowisko działa, (2) dane pobrane, (3) pytania zapisane w notatniku.
